# Activity is not price variance

At equal event intensity, independent jumps spread out while alternating jumps remain bounded. Quadratic variation still counts every squared jump.

Synthetic data only. All settings and random seeds are specified below.

```python
from pathlib import Path
import sys, json, platform
import numpy as np
import pandas as pd
# Embed figures in saved notebook outputs, even when MPLBACKEND=Agg.
%matplotlib inline
import matplotlib.pyplot as plt
from IPython.display import display
ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "pyproject.toml").exists())
sys.path.insert(0, str(ROOT / "src"))
OUT = ROOT / "results" / "synthetic" / "queues"
OUT.mkdir(parents=True, exist_ok=True)
plt.rcParams.update({"figure.figsize": (8, 4), "axes.grid": True, "grid.alpha": .2})
print({"python": platform.python_version(), "numpy": np.__version__, "pandas": pd.__version__})
```

```python
from queue_models.toy_volatility import ToyConfig,sample_terminal
rows=[]
for i,model in enumerate(("independent","alternating")):
    for rate in (10.,20.):
        x,qv,_=sample_terminal(ToyConfig(rate=rate,model=model),10.,50000,np.random.default_rng(710+i))
        rows.append(dict(model=model,rate=rate,variance=float(x.var(ddof=1)),mean_quadratic_variation=float(qv.mean())))
table=pd.DataFrame(rows);display(table)
table.to_csv(OUT/"toy_volatility.csv",index=False)
for model,g in table.groupby("model"):
    plt.plot(g.rate,g.variance,"o-",label=model)
plt.xlabel("Total event rate");plt.ylabel("Price variance at T=10");plt.legend()
plt.tight_layout();plt.savefig(OUT/"toy_volatility.png",dpi=140);plt.show()
```
